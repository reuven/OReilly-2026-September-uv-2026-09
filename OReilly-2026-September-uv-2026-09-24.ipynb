# Background

We used to describe Python as "batteries included," meaning that if you downloaded Python and installed it on your computer, you had everything that you needed to program. It included an extensive standard library that did most of the things you would do on a regular basis.

But... over time, people needed to do more than the standard library would support. And the authors of third-party projects didn't want to include their work in the Python distribution:

1. That would mean turning over copyright.
2. That would mean that new releases would come out only when there was a new release of Python.
3. Adding packages to the standard library would make the download that much larger and slower -- plus, testing all of those things to the extent that the core of Python is tested wasn't tenable.

Over time, this led to the creation of PyPI. You could then download and install any package from PyPI with the `pip` utility:

    pip install pandas

This would tell `pip` to go to PyPI, find the `pandas` project, retrieve the version that's appropriate for your platform and version of Python, and install it in `site-packages`, the directory where Python looks when you use `import`. After that, you could then say

    import pandas as pd

and it would work just great.

This meant that there were two basic sources for Python packages:

1. The Python distribution
2. PyPI, installed with `pip`

The problem is that when you download something from PyPI, it has a version number, like most software. Let's assume that you have been working on a project that needs Pandas version 2. That's fine.. until you start to work on a second project, and it needs Pandas 3.

You now have two projects on your one computer, both of which will say `import pandas as pd`, but one needs to get Pandas 2 and one neesd to get Pandas 3.

There are programming languages that allow you to install more than one version of a package. Then when you `import` (or the equivalent), you say which version you want.

Python is not one of those languages! One, and only one, version of a package can be installed with `pip` at a time. So you have to choose between Pandas 2 and Pandas 3. Or maybe ask IT to give you a separate computer for your Pandas 3 projects.

This led to a lot of complication in the Python world, and there were (and are) many different solutions to this problem, but they all revolve around something known as a "virtual environment."

This problem led to a HUGE number of conference talks! 

Finally, about 2-3 years ago, I started to hear about an alternative to `pip` known as `uv`. The big deal with `uv` was that it was super ridiculously fast, maybe 100x faster than `pip`. Maybe more. I figured that I'd switch to `uv` because it was so much faster.

How is it faster?

1. It's written in Rust, not in Python.
2. The programmers have spent a lot of time on optimization

I used `uv` for about a year and loved it. It was definitely faster than `pip`!

Then a student of mine said that I'm missing the point. That the real reason to use `uv` is not that it's faster, but that it does *everything* having to do with Python projects.

I argue that `uv` is the Python packaging super app. There have been so many different packaging-related apps and utilities over the years: `pipx`, `pyenv`, and `poetry`, and `twine`... basically, `uv` replaces them all. You only need this for :

- Installing packages
- Managing them between different projects
- Making sure that there aren't any conflicts in your project's definitions
- Creating and distributing packages
- Publishing packages to PyPI

# The real use of `uv`: Projects

You might be familiar with two different terms in Python:

- Modules, files containing Python code with a `.py` suffix that we can `import`
- Packages, directories containing modules + some metadata
- Projects, full-fledged applications that normally contain a package and also documentation, tests, metadata, etc.

`uv` is designed to work within a project. The Python world is moving pretty quickly in the direction of projects.

If you want to start working on something new, you should (in 2026+) be using a project: Create a new subdirectory with your project's name, and then start working there. `uv` basically expects this.

You can go to an existing project directory and start using `uv` there, if you want such a project. If you need (and I do this a lot), then you can create a new project with `uv init`:

- If you type `uv init` inside of a directory, that directory will now be a new Python project, and `uv` will set it up. (If you're in a non-project directory, this will be an annoying headache. If you are in an existing project directory, this won't do anything.)
- If you type `uv init NAME`, then `uv` creates a new subdirectory called `NAME` and sets it up as a project.

# Always run thing with `uv run`

It's really tempting, when you come from non-`uv` experience, and especially if your computer has one or more versions of Python on it, to just type `python myprog.py` or `python3 myprog.py`.

Don't do it! Don't give into this temptation! You really want to use `uv run python myprog.py`.

Why? You want to make sure that you're using the Python version and the packages that were installed, defined, and configured for this project.

Using `uv run python` means that you are sure the version you're running is the version specified in `.python-version` and/or in `pyproject.toml`.

# Conflicts

Remember what I said before about two different projects you're working on, A and B. Project A uses Pandas 2.x, and project B uses Pandas 3.x. When you install `pip install pandas` or `uv pip install pandas` , that gives you the latest version, aka Pandas 3. Unless there's already a version installed, in which case it won't upgrade unless you explicitly ask for it with `pip install -U pandas`.

But there's a tension here:

- Either our `site-packages` directory (which is where `pip` installs things) has Pandas 2, and project A works fine while project B cannot work, or
- Our `site-packages` directory contains Pandas 3, which means project A fails and project B works.

How has this been resolved? With venvs (virtual environments). In today's world of virtual machines, it's very tempting to think that venvs are some form of VM or cloud computing. THEY ARE NOT.

venvs are built on lies. Or on environment variables that tell Python where to look, if not in the global `site-packages`.

- Normally, when we `pip install` a package, it puts that package into `site-packages`.
- Normally, when we `import` a module/package, it looks (after elsewhere) in `site-packages`.

Venvs keep this paradigm, but specify that a *different* site-packages directory should be used. This can be set on a per-directory (per-project) basis. Similarly, you can specify a version of Python that you want to run on a per-project basis, as part of the venv's configuration.

When you go into a project directory, you "activate" the venv, meaning that you set a bunch of environment variables that tell Python where to look. When you're done there, you "deactivate" the venv, bringing you back to the global state.

This is annoying!

- Remembering to `activate` is annoying.
- The way that you `activate` is annoying.
- You have to remember to `deactivate`, which is a function, not a shell script.

People have used venvs for years. They have also complained about them for years.

The reason is not just the major Pandas 2 vs. Pandas 3 issues. It can get far more complex and subtle than that. What if you have two projects that are using slightly different versions of another project, *but* they each depend on different versions.

The good news is that venvs work. The bad news is that they are annoying. `uv` basically papers over everything having to do with venvs:

- When you install a new package, it's installed into your venv
- When you use `uv run python`, it's run with the version from your project/venv
- When you `import` in your program, assuming you ran it with `uv run`, you'll get the local `site-packages` for that venv.

# Installing packages

In the world of `uv`, we don't use `uv pip install`. Instead, you use `uv add`:

- `uv add pandas`
- `uv add rich`

This not only installs the package into your project's venv, but also adds it to `pyproject.toml`, so that anyone else running your project is guaranteed to have the right packages and the right versions.

# `uv.lock` and `pyproject.toml`

The `pyproject.toml` file is now a Python standard. It describes your project -- what version of Python, what packages does it use, etc. 

It doesn't describe what those packages depend on. So maybe `pyproject.toml` knows you need Pandas. But it won't say what Pandas needs.

`uv` introduced a new file called `uv.lock`. It lists all of the dependencies that you named, and that the packages themselves have. It also lists all of the versions for all of the operating systems, Python versions, and even hardware that are available for a give package.

`uv.lock` is basically what `uv` can use to retrieve the latest and greatest version of a package that's appropriate for the user's hardware and operating system.

NEVER EVER EVER touch `uv.lock` manually. Let `uv` manage it!

As for `pyproject.toml`, it's fine to edit the file, but be very wary of doing it without knowing the syntax or what you're doing.

Yes, you should DEFINITELY be adding, committing, and sharing `uv.lock` with others on your team?

# `uv run python` vs. `uv run` 

When you say `uv run`, you're telling `uv` to start up whatever program comes to its right as if it's an executable. When you say `uv run python XXX`, it runs Python, giving `XXX` as an argument to `python`.

If you just say `uv run XXX`, then it had better be executable! You can use the "shebang line" of `#!/usr/bin/env python` to run it, wherever it might but ..."

# Versions of packages

When you say `uv add rich`, it will take the most recent version that is compatible with your current operating system and platform. 

By default, it'll put in `pyproject.toml` the name and the minimum version number. If you install `pandas` 3, then it'll say `pandas >= 3` or the like.

You can also be oddly specific and ask for a particular version with the `==` operator:

    pandas==3.0.0

or

    pandas>=3.0.0

This is a terrible thing to do!     What's wrong? If and when Pandas 4.0 comes out, then this will say that it's compatible, which might not be the case!

So you should probably say that you have a minimum and maximum compatiblity:

    pandas>=3.0.0,<4.0.0


# `pyproject.toml` and `uv.loc` and `.venv`

- You specify, in `pyproject.toml`, what packages you're going to `import` that are not in the standard library. You'll also ideally specify a version.
- `uv` uses your `pyproject.toml` to create `uv.lock`, with all versions and all dependencies. The `uv.lock` file makes it super explicit just what files are needed for each package.
- From `uv.lock`, `uv` decides what should be in `.venv`.

Don't touch `uv.lock` or `.venv`. Just update `pyproject.toml`.

# How does `uv` deal with multiple versions?

What happens if you say `theproj>=5` in `pyproject.toml`? Will `uv` use `theproj` version 4? Or 5? or 4.5?

The answer is: The maximum allowed, so long as it doesn't break compatiblity with any other package that has been specified. The rule is always, "Get the latest and greatest that doesn't violate the specs."

Even if there are multiple versions on PyPI, `uv` might not take the latest one if that would be incompatible with your project.

# Requirements

We've seen that if a program in your project will `import` something, and if it's not in the standard library, you need to put it in `pyproject.toml`. There is no longer, when you use a venv, a global `site-packages` directory. If you `uv add` a package, it goes to the project's `site-packages` directory, under `.venv/lib/python-NN/site-packages`.

Once again, this means that if and when you share your project with someone else, they'll get the `uv.lock` and `pyproject.toml`, which will tell them what packages and versions will work.

But... this can take time. Does someone really need to install all of the packages? Yes, if they're needed for running the program.

But what about for developing the program? What if you're doing lots of automated testing with `pytest`? Then having `pytest` is needed... but not for end users who aren't going to want to modify things. They just want to run a program.

- We want and need `pytest` to be a dependency for our program
- We don't want `pytest` to be a dependency, since it'll increase the size and time needed for non-coder end users.

So... what's the solution? Dependency groups. If you want to make `pytest` a dependency, but only for developers, you can do that! Just say `uv add --dev pytest`.
